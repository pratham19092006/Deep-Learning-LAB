```python

# ------------------------------------------
# 1. Basic Example
# ------------------------------------------
print("\n===== BASIC ATTENTION =====")

sentence = ["I", "love", "deep", "learning"]
attention_scores = torch.tensor([0.1, 0.3, 0.4, 0.2])

# Softmax normalization
attention_weights = F.softmax(attention_scores, dim=0).detach().numpy()

print("Words:", sentence)
print("Attention Weights:", attention_weights)

# Plot heatmap
plt.figure(figsize=(8, 2))
sns.heatmap([attention_weights], annot=True, xticklabels=sentence, cmap="Blues")
plt.title("Basic Attention Heatmap")
plt.show()

```

```python

# ------------------------------------------
# 2. Comparison (Before vs After Softmax)
# ------------------------------------------
print("\n===== BEFORE vs AFTER SOFTMAX =====")

raw_scores = torch.tensor([2.0, 1.0, 3.0, 0.5])
softmax_scores = F.softmax(raw_scores, dim=0).detach().numpy()

plt.figure(figsize=(10, 3))

plt.subplot(1, 2, 1)
sns.heatmap([raw_scores.numpy()], annot=True, xticklabels=sentence, cmap="Reds")
plt.title("Raw Scores")

plt.subplot(1, 2, 2)
sns.heatmap([softmax_scores], annot=True, xticklabels=sentence, cmap="Blues")
plt.title("After Softmax")

plt.show()

```

```python

# ------------------------------------------
# 3. Multiple Sentences (Better Understanding)
# ------------------------------------------
print("\n===== MULTIPLE SENTENCES =====")

sentences = [
    ["I", "hate", "this", "movie"],
    ["This", "movie", "is", "amazing"]
]

attention_list = [
    torch.tensor([0.2, 0.5, 0.2, 0.1]),
    torch.tensor([0.1, 0.2, 0.2, 0.5])
]

plt.figure(figsize=(10, 4))

for i in range(len(sentences)):
    weights = F.softmax(attention_list[i], dim=0).detach().numpy()

    plt.subplot(2, 1, i+1)
    sns.heatmap([weights], annot=True, xticklabels=sentences[i], cmap="Blues")
    plt.title(f"Sentence {i+1}")

plt.tight_layout()
plt.show()

```

```python

# ------------------------------------------
# 4. Simulated Attention Matrix (Advanced)
# ------------------------------------------
print("\n===== ATTENTION MATRIX (WORD-TO-WORD) =====")

sentence = ["Deep", "learning", "is", "powerful"]

# Simulated attention matrix (4x4)
attention_matrix = torch.rand(4, 4)
attention_matrix = F.softmax(attention_matrix, dim=1).detach().numpy()

plt.figure(figsize=(6, 5))
sns.heatmap(attention_matrix, annot=True,
            xticklabels=sentence,
            yticklabels=sentence,
            cmap="viridis")

plt.title("Attention Matrix (Each word attends to others)")
plt.xlabel("Key Words")
plt.ylabel("Query Words")
plt.show()

```

```python

# ------------------------------------------
# 5. Interpretation Example
# ------------------------------------------
print("\n===== INTERPRETATION =====")

for word, weight in zip(sentence, attention_matrix[0]):
    print(f"Word '{sentence[0]}' attends to '{word}' with weight {weight:.3f}")
```
